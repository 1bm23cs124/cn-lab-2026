### Leaky Bucket Algorithm Implementation

The Leaky Bucket algorithm is a congestion control technique used in networks to shape bursty traffic into a steady stream. It works by storing incoming packets in a finite buffer (the bucket) and releasing them at a constant rate (the leak rate). If the bucket overflows, incoming packets are discarded.

### Pseudocode

```text
Initialize:
    bucket_size = 0
    capacity = Max limit of bucket
    leak_rate = Constant rate of output transmission

On Packet Arrival(packet_size):
    // Step 1: Leak packets at constant rate
    bucket_size = max(0, bucket_size - leak_rate)
    
    // Step 2: Try to buffer incoming packet
    if (bucket_size + packet_size <= capacity):
        bucket_size = bucket_size + packet_size
        Accept packet
    else:
        Discard/Drop packet (Congestion Control)
```

````python
import time
import random

class LeakyBucket:
    def __init__(self, capacity, leak_rate):
        self.capacity = capacity      # Maximum capacity of the bucket in packets/bytes
        self.leak_rate = leak_rate    # Number of packets leaked per unit of time
        self.current_water = 0        # Current content of the bucket

    def add_packet(self, packet_size):
        print(f"\nIncoming packet size: {packet_size}")

        # First, simulate the continuous leaking before processing the new packet
        # (For this discrete simulation, we leak first, then try to add)
        self.leak()

        # Check if the incoming packet can fit in the bucket
        if self.current_water + packet_size <= self.capacity:
            self.current_water += packet_size
            print(f"Packet accepted. Current bucket size: {self.current_water}/{self.capacity}")
        else:
            excess = (self.current_water + packet_size) - self.capacity
            print(f"Bucket overflow! Packet of size {packet_size} discarded. Excess: {excess}")

    def leak(self):
        if self.current_water > 0:
            leaked = min(self.current_water, self.leak_rate)
            self.current_water -= leaked
            print(f"Leaked {leaked} packets. Current bucket size: {self.current_water}/{self.capacity}")
        else:
            print("Bucket is empty. Nothing to leak.")

# --- Simulation ---
# Initialize bucket with a capacity of 10 and a leak rate of 2 packets per step
bucket = LeakyBucket(capacity=10, leak_rate=2)

# Simulate random packet arrivals over 8 steps
random.seed(42)
for step in range(1, 9):
    print(f"\n--- Time Step {step} ---")
    incoming_packet = random.randint(1, 6)
    bucket.add_packet(incoming_packet)
````
